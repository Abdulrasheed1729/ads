```python
def longestCommonPrefix(s1, s2):
    i = 0
    while i < len(s1) and i < len(s2) and s1[i] == s2[i]:
        i += 1
    return s1[:i]

seq1 = generateDnaSequence(10)
print(f"seq1: {seq1}")
seq2 = generateDnaSequence(10)
print(f"seq2: {seq2}")

print(f"Longest common prefix: {longestCommonPrefix(seq1, seq2)}")
```

```python
# One can just say s1 == s2
def match(s1, s2):
    if len(s1) != len(s2):
        return False

    for i in range(len(s1)):
        if s1[i] != s2[i]:
            return False
    return True
```

```python
match("ACGT", "ACGT")
```

```python
def reverseComplement(seq):
    complement = {"A" : "T", "C" : "G", "G" : "C", "T" : "A"}
    rev_comp = ""

    for base in seq:
        rev_comp = complement[base] + rev_comp

    return rev_comp
```

```python
seq1 = generateDnaSequence(10)
print(f"seq1: {seq1}")
revSeq1 = reverseComplement(seq1)
print(f"revSeq1: {revSeq1}")
```

```python
def readGenome(filename):
    genome = ''

    with open(filename, 'r') as f:
        for line in f:
            if not line[0] == '>':
                genome += line.rstrip()

    return genome
```

```python
genome = readGenome("lambda_virus.fa")
genome[:67]
```
